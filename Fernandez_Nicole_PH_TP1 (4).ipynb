## 1- Webscrapping

Extrae un texto de Internet de una sóla página web usando alguna de las técnicas de webscrapping vistas en clase. Busca el discurso de alguna persona quien te inspire o haya dado una charla motivadora.

No puedes repetir ni usar ni el mismo texto ni la misma web que otra persona quien ya entregó en el foro.

Abajo añade los bloques de código y texto necesarios. Explica paso a paso como descargaste el texto de tu interés.


```python
import requests
from bs4 import BeautifulSoup

url = "https://seul.ar/video/52-sebastian-campanario/"

respuesta = requests.get(url)

print("Código de respuesta:", respuesta.status_code)

soup = BeautifulSoup(respuesta.text, "html.parser")

titulo_transcripcion = soup.find(
    lambda tag: tag.name in ["h2", "h3", "h4"]
    and "Transcripción" in tag.get_text()
)

parrafos = []

if titulo_transcripcion:
    for elemento in titulo_transcripcion.find_all_next("p"):
        texto_parrafo = elemento.get_text(" ", strip=True)

        if texto_parrafo:
            parrafos.append(texto_parrafo)

texto = " ".join(parrafos)

print("Cantidad de párrafos extraídos:", len(parrafos))
print("Cantidad de caracteres:", len(texto))

print("\nPrimeros 1000 caracteres del texto extraído:\n")
print(texto[:1000])

# Para realizar el web scraping se utilizó la biblioteca `requests`, que permite realizar una solicitud HTTP a una página web y obtener su contenido HTML.
# Luego se utilizó `BeautifulSoup` para analizar la estructura HTML de la página y localizar la sección correspondiente a la transcripción.
# Una vez encontrada la sección "Transcripción", se recorrieron los párrafos que forman parte de ella y se almacenaron sus contenidos.
#Finalmente, los párrafos fueron unidos en una única variable de texto para poder utilizarlo en las siguientes etapas del trabajo.

#La fuente utilizada fue la página de la entrevista a Sebastián Campanario: https://seul.ar/video/52-sebastian-campanario/
```

## 2- Preprocesamiento

### 2.1 Muestra las stop words que tenga tu texto.

```python
import nltk
from nltk.corpus import stopwords

nltk.download("stopwords")

stop_words = set(stopwords.words("spanish"))

palabras = texto.lower().split()

stop_words_en_texto = sorted(set(palabra for palabra in palabras if palabra in stop_words))

print("Stop words encontradas en el texto:")
print(stop_words_en_texto)

print("\nCantidad de stop words diferentes:", len(stop_words_en_texto))

#Para este paso se utilizó la biblioteca `nltk`, que contiene un conjunto de palabras vacías o *stop words* en español.

#Primero se descargó el conjunto de stop words de NLTK y se obtuvo la lista correspondiente al idioma español.
#Luego, el texto extraído en el punto anterior se convirtió a minúsculas y se separó en palabras.

#Finalmente, se compararon las palabras del texto con la lista de stop words y se conservaron únicamente aquellas que aparecen en el texto.
#De esta manera, se puede observar cuáles son las stop words presentes en el texto que se utilizará para las siguientes etapas del preprocesamiento.
```

### 2.2 Quita las stop-words, caracteres extraños del texto. Utiliza el texto restante para crear una nube de palabras. Usa la biblioteca worldcloud de python.

```python
import re
import matplotlib.pyplot as plt
from wordcloud import WordCloud

texto_limpio = texto.lower()

texto_limpio = re.sub(r"[^a-záéíóúüñ\s]", " ", texto_limpio)

texto_limpio = re.sub(r"\s+", " ", texto_limpio).strip()

palabras = texto_limpio.split()

palabras_sin_stopwords = [
    palabra for palabra in palabras
    if palabra not in stop_words
]

texto_sin_stopwords = " ".join(palabras_sin_stopwords)

print("Texto original:")
print(texto[:500])

print("\nTexto después del preprocesamiento:")
print(texto_sin_stopwords[:500])

wordcloud = WordCloud(
    width=1000,
    height=500,
    background_color="white"
).generate(texto_sin_stopwords)

plt.figure(figsize=(12, 6))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.show()

# Explicación

#En este paso se realizó el preprocesamiento del texto obtenido mediante web scraping.
#Primero, se convirtió todo el texto a minúsculas para evitar que una misma palabra sea considerada diferente debido a las mayúsculas.
#Luego, se eliminaron los caracteres extraños y los signos de puntuación utilizando expresiones regulares, conservando las letras del idioma español, incluyendo las vocales acentuadas, la diéresis y la letra ñ.
#Después, el texto se separó en palabras y se eliminaron las stop words identificadas en el punto 2.1.
#Finalmente, se utilizó la biblioteca `wordcloud` para generar una nube de palabras a partir del texto limpio. En la nube, las palabras que aparecen con mayor frecuencia tienen una representación visual más grande.
```

### 2.3 Usando el texto que conseguiste en el punto 1).



#### 2.3.1 Limpia y preprocesa el texto, descomponelo en oraciones, arma una matriz documento-vocabulario.


```python
import nltk
import re
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

nltk.download("punkt")
nltk.download("punkt_tab")

oraciones = nltk.sent_tokenize(texto, language="spanish")

def limpiar_oracion(oracion):
    oracion = oracion.lower()
    oracion = re.sub(r"[^a-záéíóúüñ\s]", " ", oracion)
    oracion = re.sub(r"\s+", " ", oracion).strip()
    palabras = oracion.split()
    palabras = [palabra for palabra in palabras if palabra not in stop_words]
    return " ".join(palabras)

oraciones_limpias = [limpiar_oracion(oracion) for oracion in oraciones]

oraciones_limpias = [
    oracion for oracion in oraciones_limpias
    if oracion
]

vectorizador = CountVectorizer()

matriz_documento_vocabulario = vectorizador.fit_transform(oraciones_limpias)

vocabulario = vectorizador.get_feature_names_out()

df_documento_vocabulario = pd.DataFrame(
    matriz_documento_vocabulario.toarray(),
    columns=vocabulario
)

print("Cantidad de documentos:", len(oraciones_limpias))
print("Cantidad de palabras del vocabulario:", len(vocabulario))
print("Tamaño de la matriz:", matriz_documento_vocabulario.shape)

display(df_documento_vocabulario)
```


#### 2.3.2 Determina la longitud de tu vocabulario.

```python
longitud_vocabulario = len(vocabulario)

print("Longitud del vocabulario:", longitud_vocabulario)
```


#### 2.3.3. Aplica TF-IDF a tu matriz documento-vocabulario

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizador = TfidfVectorizer()

matriz_tfidf = tfidf_vectorizador.fit_transform(oraciones_limpias)

vocabulario_tfidf = tfidf_vectorizador.get_feature_names_out()

df_tfidf = pd.DataFrame(
    matriz_tfidf.toarray(),
    columns=vocabulario_tfidf
)

print("Cantidad de documentos:", matriz_tfidf.shape[0])
print("Cantidad de términos:", matriz_tfidf.shape[1])
print("Tamaño de la matriz TF-IDF:", matriz_tfidf.shape)

display(df_tfidf)
```

Explica el formato de tu matriz resultante: documentos, vocabulario, tamaño. Brinda tus conclusiones.

La matriz TF-IDF obtenida está formada por 465 documentos y 1725 términos diferentes del vocabulario.

Cada fila representa un documento. En este trabajo, cada documento corresponde a una oración del texto obtenido mediante web scraping. Por lo tanto, la matriz contiene 465 filas.

Cada columna representa un término diferente del vocabulario obtenido después de realizar el preprocesamiento del texto. En este caso, el vocabulario contiene 1725 términos, por lo que la matriz tiene 1725 columnas.

El tamaño de la matriz es `(465, 1725)`, es decir, 465 filas por 1725 columnas. En total, la matriz contiene 801.125 valores.

Los valores de la matriz representan los pesos TF-IDF de cada término dentro de cada documento. Un valor de 0 indica que ese término no tiene presencia en ese documento.

Como conclusión, el procesamiento permitió transformar el texto obtenido mediante web scraping en una representación numérica. Primero se realizó la limpieza del texto y la eliminación de stop words, y posteriormente se construyó la matriz documento-vocabulario. Finalmente, se aplicó TF-IDF para representar la importancia de los términos en cada documento. El resultado permite analizar el contenido del texto mediante técnicas de procesamiento del lenguaje natural.
